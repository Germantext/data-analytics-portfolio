## 2. Gasto promedio entre usuarios convertidos (página A vs B)


```python

# Gasto por versión
gasto_A = df[(df['landing'] == 'A') & (df['converted'] == 1)]['gasto']
gasto_B = df[(df['landing'] == 'B') & (df['converted'] == 1)]['gasto']

# Verificar cantidad de datos que tiene cada grupo
len(gasto_A), len(gasto_B)
```

### 2.1 Prueba de Levene y prueba t de Student

Hipótesis:
- Hipótesis nula (H₀): El gasto promedio de los usuarios convertidos es igual en la página A y en la página B.
- Hipótesis alternativa (H₁): El gasto promedio de los usuarios convertidos es diferente entre la página A y la página B.


```python
# Aplicar prueba de Levene (verificar igualdad de varianzas)
l_stat, p_value_var = levene(gasto_A, gasto_B)

alpha = 0.05
equal_var = p_value_var >= alpha

# Aplicar prueba t
t_stat, p_value = ttest_ind(gasto_A, gasto_B, equal_var=equal_var)

# Visualizar resultados
print(f"Estadístico de Levene: {l_stat}")
print(f"Valor p (Levene): {p_value_var}")
print(f"¿Varianzas iguales? {equal_var}")
print(f"\nEstadístico t: {t_stat}")
print(f"Valor p: {p_value}")
```

```python
print(f"Gasto promedio página A: {gasto_A.mean():.2f}")
print(f"Gasto promedio página B: {gasto_B.mean():.2f}")
```

### 📝 Conclusión e interpretación

Decisión:
El valor p obtenido (3.63e-21) es mucho menor que el umbral de significancia (α = 0.05), 
por lo que rechazamos la hipótesis nula. Existe evidencia estadística de que el gasto 
promedio de los usuarios convertidos difiere entre la página A y la página B.

Interpretación de negocio:
Los usuarios que convirtieron en la página B gastaron en promedio 68.75 USD, mientras que 
los de la página A gastaron en promedio 61.09 USD, una diferencia de 7.66 USD por usuario 
convertido. La prueba de Levene indicó que las varianzas entre ambos grupos no son 
iguales (p = 6.88e-08), por lo que se utilizó la corrección de Welch para la prueba t, 
garantizando un resultado confiable.

Este resultado se basa en una muestra de usuarios convertidos durante el periodo del 
experimento y no evalúa por sí solo si la diferencia de 7.66 USD es suficiente para 
justificar la implementación definitiva de la página B, ya que esto dependería también 
de otros factores como costos de implementación y el comportamiento en la tasa de 
conversión (que se evalúa en el siguiente paso).



## 3. Tasa de conversión entre la página A y B


3.1. Prueba z de proporciones

Hipótesis:
- Hipótesis nula (H₀): La tasa de conversión es igual en la página A y en la página B.
- Hipótesis alternativa (H₁): La tasa de conversión es diferente entre la página A y la página B.

```python

# Número de usuarios convertidos por página
conversiones = df.groupby('landing')['converted'].sum()

# Total de usuarios por página
totales = df.groupby('landing')['converted'].count()

print("Usuarios convertidos por página:\n", conversiones)
print("\nTotal de usuarios por página:\n", totales)
```

```python

# Aplicar prueba
exitos = [conversiones['A'], conversiones['B']]
observaciones = [totales['A'], totales['B']]

z_stat, p_value = proportions_ztest(exitos, observaciones)

# Visualizar resultados
print(f"Estadístico z: {z_stat}")
print(f"Valor p: {p_value}")

alpha = 0.05
if p_value < alpha:
    print("Rechazamos la hipótesis nula: hay evidencia de una diferencia.")
else:
    print("No rechazamos la hipótesis nula: no hay evidencia suficiente de una diferencia.")
```

```python
tasa_A = conversiones['A'] / totales['A']
tasa_B = conversiones['B'] / totales['B']

print(f"Tasa de conversión página A: {tasa_A:.2%}")
print(f"Tasa de conversión página B: {tasa_B:.2%}")
print(f"Diferencia: {tasa_B - tasa_A:.2%}")
```

### 📝 Conclusión e interpretación

Decisión:
El valor p obtenido (3.76e-22) es mucho menor que el umbral de significancia (α = 0.05), 
por lo que rechazamos la hipótesis nula. Existe evidencia estadística de que la tasa de 
conversión difiere entre la página A y la página B.

Interpretación de negocio:
La página B tiene una tasa de conversión de 15.96%, mientras que la página A tiene una 
tasa de conversión de 12.57%, una diferencia de 3.38 puntos porcentuales a favor de B. 
Esto significa que, además de generar un mayor gasto promedio por usuario convertido 
(visto en el paso anterior), la página B también logra convertir a una mayor proporción 
de los usuarios expuestos a ella.

Este resultado se basa en una muestra y no evalúa por sí solo si la magnitud de la 
diferencia es suficiente para justificar el cambio, considerando también el costo de 
implementación. Sin embargo, combinado con el resultado del gasto promedio, la evidencia 
apunta consistentemente hacia la página B como la versión de mayor desempeño.

## 4. Relación entre la fuente de tráfico y la conversión


### 4.1. Prueba chi-cuadrada de independencia

Hipótesis:
- Hipótesis nula (H₀): La conversión es independiente de la fuente de tráfico.
- Hipótesis alternativa (H₁): La conversión depende de la fuente de tráfico.

```python
# Tabla de contingencia
tabla_trafico = pd.crosstab(df['traffic_source'], df['converted'])
print(tabla_trafico)
```

```python
# Aplicar prueba
chi2_stat, p_value, dof, expected = chi2_contingency(tabla_trafico)

print(f"Estadístico chi-cuadrado: {chi2_stat:.3f}")
print(f"Valor p: {p_value:.3f}")
print(f"Grados de libertad: {dof}")

alpha = 0.05
if p_value < alpha:
    print("\nRechazamos la hipótesis nula: hay evidencia de asociación entre las variables.")
else:
    print("\nNo rechazamos la hipótesis nula: no hay evidencia suficiente de asociación entre las variables.")

```

```python
# Tabla de contingencia normalizada (tasas de conversión por canal)
tabla_trafico_norm = pd.crosstab(df['traffic_source'], df['converted'], normalize='index') * 100
print(tabla_trafico_norm)
```

### 📝 Conclusión e interpretación

Decisión:
El valor p obtenido (0.034) es menor que el umbral de significancia (α = 0.05), por lo 
que rechazamos la hipótesis nula. Existe evidencia estadística de que la conversión está 
asociada a la fuente de tráfico.

Interpretación de negocio:
En términos absolutos, Organic es el canal con mayor volumen de conversiones (2,480), 
seguido de Ads (1,759), Email (918) y Referral (549) — esto refleja principalmente el 
volumen de tráfico de cada canal, no necesariamente su efectividad relativa.

Al observar las tasas de conversión (normalizadas), las diferencias entre canales son 
pequeñas: Email convierte al 14.99%, Ads al 14.74%, Referral al 13.88% y Organic al 
13.79%. Aunque la prueba chi-cuadrada detecta que estas diferencias no se deben al azar 
(p = 0.034), la magnitud de la diferencia entre el canal más y menos efectivo es de 
apenas 1.2 puntos porcentuales.

Esto es un ejemplo de un resultado estadísticamente significativo pero de bajo impacto 
práctico: la asociación es real, pero no sugiere que el negocio deba reasignar 
drásticamente su inversión en tráfico solo con base en este hallazgo. Se recomienda 
evaluar el costo por adquisición de cada canal antes de tomar decisiones de inversión.


## 5. Relación entre el tipo de usuario y la conversión

Se analiza si existe una **asociación estadísticamente significativa** entre el **tipo de usuario** (`user_type`) y la **conversión** (`converted`), entendiendo que un usuario recurrente puede haber visitado antes sin necesariamente convertirse en cliente en esta ocasión.

El objetivo es identificar qué perfiles muestran mayor probabilidad de conversión dentro del contexto analizado.

5.1. Prueba chi-cuadrada de independencia

Hipótesis:
- Hipótesis nula (H₀): La conversión es independiente del tipo de usuario.
- Hipótesis alternativa (H₁): La conversión depende del tipo de usuario.

```python
# Tabla de contingencia
tabla_usuario = pd.crosstab(df['user_type'], df['converted'])
print(tabla_usuario)
```

```python
# Tabla de contingencia normalizada (tasas de conversión por tipo de usuario)
tabla_usuario_norm = pd.crosstab(df['user_type'], df['converted'], normalize='index') * 100
print(tabla_usuario_norm)
```

```python
# Aplicar prueba
chi2_stat, p_value, dof, expected = chi2_contingency(tabla_usuario)

print(f"Estadístico chi-cuadrado: {chi2_stat:.3f}")
print(f"Valor p: {p_value:.3f}")
print(f"Grados de libertad: {dof}")

alpha = 0.05
if p_value < alpha:
    print("\nRechazamos la hipótesis nula: hay evidencia de asociación entre las variables.")
else:
    print("\nNo rechazamos la hipótesis nula: no hay evidencia suficiente de asociación entre las variables.")

```

### 📝 Conclusión e interpretación

Decisión:
El valor p obtenido (0.474) es mayor que el umbral de significancia (α = 0.05), por lo 
que no rechazamos la hipótesis nula. No hay evidencia estadística suficiente para 
afirmar que la conversión está asociada al tipo de usuario.

Interpretación de negocio:
Las tasas de conversión son prácticamente idénticas entre ambos perfiles: los usuarios 
nuevos convierten al 14.36% y los recurrentes al 14.09%, una diferencia de apenas 0.27 
puntos porcentuales. El estadístico chi-cuadrado (0.513) y el valor p (0.474) confirman 
que esta diferencia es consistente con lo esperado por azar.

A diferencia de la fuente de tráfico (donde sí se detectó una asociación estadística), 
el tipo de usuario no parece ser un factor relevante para explicar diferencias en la 
conversión dentro de este experimento. Esto sugiere que no es necesario diseñar 
estrategias diferenciadas por tipo de usuario (nuevo vs. recurrente) a partir de este 
hallazgo, y que otros factores (como el dispositivo o el canal de tráfico) tienen 
mayor peso explicativo en el comportamiento de conversión.

### Insight ejecutivo basado en el experimento A/B

**7.1 Comparación de página (A vs B)**

Gasto promedio por usuario convertido:
- La página B genera un gasto promedio de 68.75 USD por usuario convertido, frente a 61.09 USD en la página A.
- La diferencia (7.66 USD) es estadísticamente significativa (prueba t de Welch, p = 3.63e-21).
- Interpretación: entre quienes convirtieron, los usuarios de la página B gastaron más en promedio que los de la página A.

Tasa de conversión:
- La página B tiene una tasa de conversión de 15.96%, frente a 12.57% en la página A.
- La diferencia (3.38 puntos porcentuales) es estadísticamente significativa (prueba z de proporciones, p = 3.76e-22).
- Interpretación: la página B convierte a una proporción mayor de usuarios expuestos, y quienes convierten en ella también gastan más en promedio. Ambos indicadores apuntan en la misma dirección, aunque se midieron por separado y no como un único indicador de ingreso por usuario expuesto.

**7.2 Segmentación por fuente de tráfico**
- Existe una asociación estadísticamente significativa entre la fuente de tráfico y la conversión (chi-cuadrado, p = 0.034), pero las tasas de conversión por canal son similares entre sí: Email (14.99%), Ads (14.74%), Referral (13.88%) y Organic (13.79%), con una diferencia máxima de apenas 1.2 puntos porcentuales.
- Interpretación: aunque la diferencia es estadísticamente significativa, su magnitud es pequeña. No se recomienda reasignar drásticamente la inversión en tráfico solo con base en este resultado; conviene evaluarlo junto con el costo de adquisición de cada canal.

**7.3 Segmentación por tipo de usuario**
- No se encontró evidencia estadísticamente significativa de asociación entre el tipo de usuario y la conversión (chi-cuadrado, p = 0.474). Los usuarios nuevos convierten al 14.36% y los recurrentes al 14.09%, una diferencia mínima y no significativa.
- Interpretación: en esta muestra no hay evidencia de que el tipo de usuario explique diferencias en conversión; no se justifica una estrategia diferenciada por este segmento a partir de este hallazgo.

Las visualizaciones usadas respaldan los resultados estadísticos de los pasos anteriores.

**7.4 Recomendaciones de negocio**
- Los resultados respaldan avanzar con la página B como versión de referencia, dado que supera de forma estadísticamente significativa a la página A tanto en tasa de conversión como en gasto promedio por usuario convertido. La recomendación queda sujeta a validar el costo de implementación y a confirmar que la asignación de usuarios entre A y B fue verdaderamente aleatoria.
- Antes de reasignar presupuesto entre canales de tráfico, evaluar el costo de adquisición por canal; la diferencia en efectividad entre fuentes es estadísticamente significativa pero de magnitud pequeña (~1 punto porcentual), por lo que otros factores (como el CAC) deberían pesar más en esa decisión.
- No se recomienda segmentar la estrategia por tipo de usuario (nuevo vs. recurrente), ya que no se encontró diferencia significativa en su comportamiento de conversión.

**7.5 Limitaciones**
- El gasto se comparó únicamente entre usuarios que convirtieron, es decir, "gasto promedio por usuario convertido", no "gasto promedio por usuario expuesto". Ambas son preguntas distintas: esta última requeriría calcular el ingreso total generado por cada grupo dividido entre el total de usuarios expuestos, cálculo que no se realizó en este análisis y que sería un buen siguiente paso.
- No se evaluó el tamaño del efecto, el intervalo de confianza de las diferencias, ni posibles desbalances entre A y B por región, dispositivo o canal.
